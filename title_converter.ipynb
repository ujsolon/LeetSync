```python
def format_title(title: str) -> str:
    # Split the input title into year and rest of the title
    number, rest_of_title = title.split('. ', 1)
    # Convert the rest of the title to lowercase and replace spaces with hyphens
    formatted_title = rest_of_title.lower().replace(' ', '-')
    # Construct the final output with the "m" indicator for manual input
    return f"{number}-m-{formatted_title}"

# Example usage
title = "6. Zigzag Conversion"
formatted = format_title(title)
print(formatted)  # Output: 2041-m-the-latest-login-in-2020
```

```python
def format_url(url: str) -> str:
    # Extract the part after "/problems/"
    problem_name = url.split("/problems/")[1].split("/")[0]
    # Convert to lowercase and replace spaces (if any) with hyphens
    formatted_title = problem_name.lower().replace(' ', '-')
    # Add a placeholder year (e.g., "2024") and the "m" indicator for manual input
    return f"m-{formatted_title}"

# Example usage
url = "https://leetcode.com/problems/longest-common-prefix/description/?envType=study-plan-v2&envId=top-interview-150"
formatted = format_url(url)
print(formatted)  # Output: 2024-m-longest-common-prefix
```
